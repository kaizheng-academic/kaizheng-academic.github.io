# TRACE E1 Static Notebook Results

This notebook summarizes reproducibility outputs generated from `TRACE_e1/outputs/run_demo_001`.

- Focus: executed outputs for summary metrics and quality gates
- Source assets: `documentation/notebooks/assets/*`


```python
from pathlib import Path
import csv
import json

ROOT = Path('assets')
ROOT.exists(), ROOT
```

```python
summary_file = ROOT / 'summary' / 'trace_e1_summary.csv'
rows = list(csv.DictReader(summary_file.open()))
print('summary rows:', len(rows))
for r in rows[:5]:
    print(r)
```

```python
diff_file = ROOT / 'summary' / 'trace_e1_diff_vs_golden.csv'
diff_rows = list(csv.DictReader(diff_file.open()))
print('diff rows:', len(diff_rows))
for r in diff_rows[:5]:
    print(r)
```

```python
repro = json.loads((ROOT / 'checks' / 'repro_check.json').read_text())
gate = json.loads((ROOT / 'checks' / 'gate_report.json').read_text())
manifest = json.loads((ROOT / 'checks' / 'run_manifest.json').read_text())
print('repro_check keys:', sorted(repro.keys()))
print('gate_report keys:', sorted(gate.keys()))
print('run_manifest keys:', sorted(manifest.keys()))
```

```python
pred_file = ROOT / 'raw' / 'trace_predictions.csv'
pred_rows = list(csv.DictReader(pred_file.open()))
print('prediction rows:', len(pred_rows))
print('first prediction record:')
print(pred_rows[0] if pred_rows else 'N/A')
```
